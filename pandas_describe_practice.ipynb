## 2. `df.describe()` - the statistical summary

One call, and every **numeric** column gets 8 numbers.

```python
df.describe()
```

### The same thing for a single column (a Series)

```python
df["AveragePrice"].describe()
```

### What each row means

| row | meaning |
|---|---|
| `count` | how many **non-empty** values the column has |
| `mean` | the average |
| `std` | standard deviation - how spread out the values are around the mean |
| `min` | the smallest value |
| `25%` | **Q1**: 25% of the values are below this |
| `50%` | the **median**: half the values are below, half above |
| `75%` | **Q3**: 75% of the values are below this |
| `max` | the largest value |

Read as a sentence for `AveragePrice`: *18249 prices, averaging 1.41 dollars, the cheapest
0.44 and the dearest 3.25; half of them are under 1.37, and the middle 50% sit between
1.10 and 1.66.*

`count` is the quiet one - if it is smaller than the number of rows, that column has
**missing values**.

## 3. Check describe() by hand

The same numbers, computed one by one. `std` uses **n - 1** (the sample version), exactly
like the `st_dev` we wrote in the functions lab.

```python
price = df["AveragePrice"]

print("count :", price.count())
print("mean  :", round(price.mean(), 6))
print("std   :", round(price.std(), 6))        # ddof=1 by default -> divides by n-1
print("min   :", price.min())
print("25%   :", price.quantile(0.25))
print("50%   :", price.median(), "= quantile(0.5):", price.quantile(0.50))
print("75%   :", price.quantile(0.75))
print("max   :", price.max())

# numpy's std divides by n instead, so it differs slightly:
import numpy as np
print()
print("pandas std (n-1):", round(price.std(), 6))
print("numpy  std (n)  :", round(np.std(price.values), 6))
```

---
## 4. describe() on the text columns

By default `describe()` ignores text. Ask for it and the rows change completely.

```python
# include="object" also works, but pandas 3 warns about it:
# text columns are now the "str" dtype
df.describe(include="str")
```

| row | meaning |
|---|---|
| `count` | how many values |
| `unique` | how many **different** values |
| `top` | the most frequent value |
| `freq` | how often the top value appears |

`include="all"` puts numbers and text in one table (with `NaN` where a statistic
does not apply).

```python
df.describe(include="all")
```

## 5. Two useful variations

```python
# your own percentiles instead of 25/50/75
df["AveragePrice"].describe(percentiles=[0.1, 0.5, 0.9])
```

```python
# describe() AFTER filtering - the summary follows the rows you kept
organic = df[df["type"] == "organic"]
conventional = df[df["type"] == "conventional"]

print("organic     :", len(organic), "rows, mean price", round(organic["AveragePrice"].mean(), 3))
print("conventional:", len(conventional), "rows, mean price", round(conventional["AveragePrice"].mean(), 3))
print()
organic["AveragePrice"].describe()
```

```python
# the same comparison in one line
df.groupby("type")["AveragePrice"].describe()
```

---
## Remember

- `df.describe()` -> numeric columns only; `include="object"` -> text; `include="all"` -> both
- `50%` **is** the median, and `25%`/`75%` mark the middle half of the data
- pandas `std` divides by **n - 1**, numpy's `np.std` divides by **n**
- `count` lower than `df.shape[0]` is your first hint of missing values
