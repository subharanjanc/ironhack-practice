# `reduce()` - a simple program, step by step

`reduce` takes a list and **boils it down to ONE value**.

It works two items at a time:

```
[1, 2, 3, 4]
 1 + 2 = 3
     3 + 3 = 6
         6 + 4 = 10   ->  10
```

Unlike `map` and `filter`, it is **not** built in - it must be imported.

---
## Step 1: the same job with a plain loop (what we already know)

````python
numbers = [1, 2, 3, 4]

total = 0
for n in numbers:
    total = total + n

print(total)
````

---
## Step 2: the same job with `reduce`

The function gets **two** parameters: `x` is the result so far, `y` is the next item.

````python
from functools import reduce

def add(x, y):
    return x + y

print(reduce(add, numbers))
````

---
## Step 3: watch it work

Printing inside the function shows the two values it receives each time.

````python
def add_loud(x, y):
    print(f"  {x} + {y} = {x + y}")
    return x + y

result = reduce(add_loud, numbers)
print("result:", result)
````

---
## Step 4: the same thing with a lambda

A `lambda` is the short form - no `def`, no `return`.

````python
print(reduce(lambda x, y: x + y, numbers))      # 10
print(reduce(lambda x, y: x * y, numbers))      # 24  (1*2*3*4)
````

---
## Step 5: a starting value

A third argument says where to start. It also makes an **empty list** safe -
without it, `reduce([])` raises `TypeError`.

````python
print(reduce(lambda x, y: x + y, numbers, 100))   # 100 + 1 + 2 + 3 + 4
print(reduce(lambda x, y: x + y, [], 0))          # 0, no error
````

---
## Step 6: reduce works on strings too

This is exactly what **Challenge 3** of the lab asks for - two strings joined with a space -
just with our own words instead of the book.

````python
def concat_space(a, b):
    '''
    Input: two strings
    Output: one string with a space between them

    Example: 'John', 'Smith'  ->  'John Smith'
    '''
    return a + " " + b


print(concat_space("John", "Smith"))

words = ["python", "is", "fun", "with", "Ironhack"]
print(reduce(concat_space, words))
````

---
## Step 7: useful without a lambda at all

`reduce` can pick a winner as well as add things up.

````python
marks = [45, 78, 92, 33]

print(reduce(lambda x, y: x if x > y else y, marks))   # biggest: 92
print(reduce(lambda x, y: x if x < y else y, marks))   # smallest: 33
````

---
## Remember

- `reduce(function, list)` -> **one value**
- the function always takes **two** arguments: *result so far* and *next item*
- `from functools import reduce` first
- an optional third argument is the **starting value**
