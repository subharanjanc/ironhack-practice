## Task 1: store the age, and use it in `say_hello`

Two changes:
- `__init__` takes a second parameter `age` and stores it as `self.age`
- `say_hello` prints the age as well

`self.age` is a **number**, so it has to be turned into text with `str()` before joining it
with `+`. An f-string does that conversion for you.

```python
class Person:
    def __init__(self, name, age):
        self.name = name          # instance attribute 1
        self.age = age            # instance attribute 2 (new)

    def say_hello(self):
        # f-string: whatever is inside {} is converted to text automatically
        print(f"hello, my name is {self.name} and I am {self.age} years old")


p = Person("Subha", 30)
p.say_hello()

# The same line written without an f-string (str() is needed around the number)
print("hello, my name is " + p.name + " and I am " + str(p.age) + " years old")

p = Person("Pinky", 25)
p.say_hello()

# The same line written without an f-string (str() is needed around the number)
print("hello, my name is " + p.name + " and I am " + str(p.age) + " years old")
```

## Bonus 1: a method `get_older()`

The method takes no extra input - it only needs `self`, because the age it changes is the
age **of this object**. Every object keeps its own `age`.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def say_hello(self):
        print(f"hello, my name is {self.name} and I am {self.age} years old")

    def get_older(self):
        # self.age = self.age + 1 changes the attribute stored inside THIS object
        self.age = self.age + 1


p = Person("Subha", 30)
print("age before:", p.age)

p.get_older()
print("age after :", p.age)

p.say_hello()
```

## Bonus 2: every time the user gets older, call `say_hello`

One method can call another one on the same object - `self.say_hello()`.
So `get_older()` first adds a year, then greets with the new age.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def say_hello(self):
        print(f"hello, my name is {self.name} and I am {self.age} years old")

    def get_older(self):
        self.age = self.age + 1
        self.say_hello()          # a method calling another method of the same object


p = Person("Subha", 30)
p.say_hello()

p.get_older()   # 31 - and it greets on its own
p.get_older()   # 32
p.get_older()   # 33
```

## Two objects, two separate ages

This is the point of instance attributes: each object carries its own data, and `self` is
how a method knows which object it is working on.

```python
subha = Person("Subha", 30)
roberto = Person("Roberto", 25)

subha.get_older()      # only Subha has a birthday here

print()
print("subha.age  :", subha.age)
print("roberto.age:", roberto.age)

# A birthday for everyone, using a loop over the objects
print()
for person in [subha, roberto]:
    person.get_older()
```

## What to remember

| Term | Meaning |
|---|---|
| `class Person:` | the blueprint |
| `p = Person("Subha", 30)` | an **instance** (one real object built from the blueprint) |
| `__init__` | runs automatically when the object is created |
| `self` | the object itself - Python passes it in, you never write it in the call |
| `self.age` | an **instance attribute**: data that belongs to that one object |
| `self.say_hello()` | a method calling another method on the same object |

`p.say_hello()` is really `Person.say_hello(p)` - that is why every method's first
parameter is `self`.
