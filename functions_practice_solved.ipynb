## Part A (09:55): functions and function parameters

### Iteration 1
Create a function called `display_sum()` that:
- Takes two numbers as parameters (e.g., `num1` and `num2`)
- Calculates the sum of those two numbers
- Displays (prints) the total to the screen
- **Do not use `return` yet**

Call the function at least 3 times with different numbers.
For example, calling `display_sum(10, 5)` should display `15`.

```python
# Part A - Iteration 1: display_sum()
# Adds two numbers and prints the total (no return yet)
def display_sum(num1, num2):
    total = num1 + num2
    print(total)

# Call it 3 times with different numbers
display_sum(10, 5)
display_sum(3, 7)
display_sum(100, 250)
```

### Iteration 2
Create a function `display_mult()` to calculate the multiplication of 2 numbers.

```python
# Part A - Iteration 2: display_mult()
# Multiplies two numbers and prints the result (no return)
def display_mult(num1, num2):
    product = num1 * num2
    print(product)

display_mult(4, 5)
display_mult(7, 3)
display_mult(12, 10)
```

### Bonus
- Test what happens if you invoke the function passing only one value, e.g. `display_sum(10)`
- Test your function with negative and/or decimal numbers
- Research and practice: "Python type hints"
- Research and practice: "default parameters"

*Time: 15 min*

```python
# Part A - Bonus

# 1) Calling with only one value: Python needs both parameters, so it raises a TypeError
try:
    display_sum(10)
except TypeError as error:
    print("Error:", error)

# 2) Negative and decimal numbers work too
display_sum(-8, 3)
display_sum(2.5, 1.25)
display_mult(-3, 1.5)

# 3) Type hints (: float, -> None) describe the expected types; Python does not enforce them
# 4) A default parameter (num2 = 0) is used when that value is not passed
def display_sum_typed(num1: float, num2: float = 0) -> None:
    print(num1 + num2)

display_sum_typed(10)      # num2 uses its default 0
display_sum_typed(10, 5)
```

## Part B (11:04): functions, parameters, and return values

### Iteration 1
Create a function called `calculate_difference()` that:
- Takes two numbers as parameters (`num1` and `num2`)
- Calculates the difference (`num1 - num2`)
- **Returns** the result (important: **do not print inside the function**, it needs to return the result)

Call the function at least 3 times with different numbers and **print the returned value**.

```python
# Part B - Iteration 1: calculate_difference()
# Returns num1 - num2 (no print inside the function)
def calculate_difference(num1, num2):
    difference = num1 - num2
    return difference

# Call it 3 times and print the returned value outside the function
print(calculate_difference(10, 3))
print(calculate_difference(5, 12))
print(calculate_difference(100, 45))
```

### Iteration 2
Create a function called `calculate_division()` that:
- Takes two numbers as parameters
- Returns the result of dividing the first number by the second

```python
# Part B - Iteration 2: calculate_division()
# Returns the first number divided by the second
def calculate_division(num1, num2):
    result = num1 / num2
    return result

print(calculate_division(10, 2))
print(calculate_division(7, 2))
print(calculate_division(1, 3))
```

### Iteration 3
Create a function called `calculate_total()` that:
- Takes a list of numbers. Example: `[5, 10, 15]`
- Returns the sum of all of them. Example: `30`

```python
# Part B - Iteration 3: calculate_total()
# Takes a list of numbers and returns the sum of all of them
def calculate_total(numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(calculate_total([5, 10, 15]))   # example from Luis: 30
print(calculate_total([1, 2, 3, 4]))
print(calculate_total([]))            # an empty list sums to 0
```

### Bonus

```python
# Part B - Bonus: default parameters, type hints and a docstring

def calculate_power(base: float, exponent: int = 2) -> float:
    """Return base raised to exponent.

    exponent defaults to 2, so calculate_power(5) gives 25.
    """
    return base ** exponent

print(calculate_power(5))        # uses the default exponent 2
print(calculate_power(2, 10))    # passes exponent explicitly

# The docstring is stored on the function and can be read back
print(calculate_power.__doc__)
```
