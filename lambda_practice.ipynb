A **lambda** is a small one-line function without a `def`:

```python
name = lambda parameters: expression   # the expression is returned automatically
```

---
## Iteration 1
Create a lambda function called `square` that:
- Takes one number as input
- Returns its square

Test it with different numbers.
e.g., `square(5)` should return `25`; `square(-3)` should return `9`

````python
# Iteration 1: square
# One input x, returns x * x (no return keyword needed in a lambda)
square = lambda x: x ** 2

print(square(5))     # expected 25
print(square(-3))    # expected 9
print(square(0))
print(square(1.5))
````

## Iteration 2
Create a lambda function called `multiply` that:
- Takes two numbers as input
- Returns their product

Test it with positive, negative, and decimal numbers

````python
# Iteration 2: multiply
# Two inputs separated by a comma, returns their product
multiply = lambda a, b: a * b

print(multiply(4, 5))      # positive: 20
print(multiply(-3, 7))     # negative: -21
print(multiply(-2, -6))    # two negatives make a positive: 12
print(multiply(2.5, 4))    # decimal: 10.0
````

## Bonus 1
Create a lambda function called `get_last_char`. This function should take a string as an argument and return only the last character of the string.

````python
# Bonus 1: get_last_char
# Index -1 means "the last position" in a string
get_last_char = lambda text: text[-1]

print(get_last_char("hello"))       # o
print(get_last_char("Ironhack"))    # k
print(get_last_char("A"))           # a one-letter string returns that letter

# Note: text[-1] raises an IndexError on an empty string "".
# The slice text[-1:] returns "" instead, which is safer:
get_last_char_safe = lambda text: text[-1:]
print(repr(get_last_char_safe("")))
````

## Bonus 2
Create a lambda function called `count_vowels` that takes a string as input, and returns the number of vowels in the string.
- e.g., `count_vowels("hello")` should return `2`
- e.g., `count_vowels("Python is cool")` should return `4`

````python
# Bonus 2: count_vowels
# .lower() makes capitals count too; sum() adds 1 for every character that is a vowel
count_vowels = lambda text: sum(1 for char in text.lower() if char in "aeiou")

print(count_vowels("hello"))            # expected 2
print(count_vowels("Python is cool"))   # expected 4
print(count_vowels("AEIOU"))            # capitals: 5
print(count_vowels("rhythm"))           # no vowels: 0
````

## lambda vs def: the same function written both ways

````python
# The lambda version
square_lambda = lambda x: x ** 2

# The equivalent def version: needs a name line, a body, and an explicit return
def square_def(x):
    return x ** 2

print(square_lambda(6), square_def(6))   # both give 36
````
