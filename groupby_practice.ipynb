The gist loads the data with `sns.load_dataset("titanic")`. seaborn is not installed here, so we
read the same file (already downloaded to `data/titanic.csv`) with pandas.

```python
import pandas as pd

df = pd.read_csv("data/titanic.csv")
print(df.shape)
df.head()
```

---
## Iteration 1
Find the average `age` for each passenger `class`.

```python
# group the rows by class, take only the age column, and average it per group
df.groupby("class")["age"].mean()
```

---
## Iteration 2
Find the average `fare` and `survived` rate grouped by `class`, using `agg()` to get both in one call.

```python
# agg() takes a dictionary: {column: calculation}, so both results come from one groupby
df.groupby("class").agg({"fare": "mean", "survived": "mean"})
```

---
## Iteration 3
Group by `class` and `sex` together, and find the survival rate (`survived` mean) for each
combination. Use `as_index=False`.

```python
# a LIST of columns groups by every combination of class and sex;
# as_index=False keeps class and sex as normal columns instead of a two-level index
df.groupby(["class", "sex"], as_index=False)["survived"].mean()
```

---
## Bonus 1
For each `class`, find the min, max, and mean `fare`, all in one `agg()` call.

```python
# a LIST inside agg() runs several calculations on the same column in one call
df.groupby("class")["fare"].agg(["min", "max", "mean"])
```

## Bonus 2
Group by `embark_town` and find the average `fare`. Then identify which `embark_town` has the
highest average fare (without manually scanning the output).

```python
# average fare per embarkation town
avg_fare_by_town = df.groupby("embark_town")["fare"].mean()
print(avg_fare_by_town)
print()

# idxmax() returns the INDEX LABEL of the largest value - here, the town name -
# so we never have to read the table by eye
top_town = avg_fare_by_town.idxmax()
print("highest average fare:", top_town, "->", round(avg_fare_by_town.max(), 2))
```
