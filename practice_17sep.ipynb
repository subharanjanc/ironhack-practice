```python


print("numpy version:", np.__version__)

a = np.array([1, 2, 3, 4, 5])
print("array   :", a)
print("type    :", type(a))

print("a * 2   :", a * 2)          # every item doubled - no loop, no map
print("sum     :", a.sum(), "| mean:", a.mean(), "| std:", a.std())
print("sqrt    :", np.sqrt(a))

m = np.array([[1, 2], [3, 4]])     # a 2x2 matrix
print("matrix  :"); print(m)
print("shape   :", m.shape)
print("transpose:"); print(m.T)
```

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'], name='marks')
print(s)
```

```python
print(s[::-1])        # slice with step -1 — same trick as on a Python list
print(s.iloc[::-1])   # the explicit, recommended way: by position
```

```python
import pandas as pd

df = pd.read_csv("data/avocado_kaggle.csv")

print(df.shape)        # (18249, 14)
df.head()
```

```python
df.info()                        # column names, types, missing values
df.describe()                    # count/mean/std/min/max of the numeric columns

prices = df["AveragePrice"]      # one column = a pandas Series
print(prices.mean())             # 1.406
print(prices.min(), prices.max())

print(df["type"].unique())       # ['conventional' 'organic']
print(df["region"].nunique())    # 54
print(df.groupby("type")["AveragePrice"].mean())
```

```python
df[df["region"].isin(["Albany", "Boston", "Chicago"])]   # 1014 rows
df[df["AveragePrice"].between(1.0, 1.5)]                 # 8541 rows
df[df["region"].str.contains("New")]                     # 1349 rows
# → ['NewOrleansMobile', 'NewYork', 'NorthernNewEngland', 'WestTexNewMexico']
```
